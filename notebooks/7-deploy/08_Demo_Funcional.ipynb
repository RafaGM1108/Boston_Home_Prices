## 📋 Descripción de la demo

La demo consiste en un formulario web interactivo con los siguientes elementos:

### Entradas (13 features del dataset)

| Feature | Tipo de input | Rango | Descripción |
|---------|--------------|-------|-------------|
| `crim` | number_input | 0 - 90 | Tasa de criminalidad per cápita |
| `zn` | number_input | 0 - 100 | Proporción terreno residencial (%) |
| `indus` | number_input | 0 - 28 | Proporción zona industrial (%) |
| `chas` | selectbox | 0, 1 | Colinda con río Charles (Sí/No) |
| `nox` | number_input | 0.38 - 0.87 | Concentración de contaminantes |
| `rm` | number_input | 3 - 9 | Número promedio de habitaciones |
| `age` | number_input | 0 - 100 | Antigüedad de viviendas (%) |
| `dis` | number_input | 1 - 12 | Distancia a centros de empleo |
| `rad` | selectbox | 1-8, 24 | Índice accesibilidad autopistas |
| `tax` | number_input | 180 - 720 | Tasa de impuesto a la propiedad |
| `ptratio` | number_input | 12 - 22 | Ratio alumnos/profesor |
| `black` | number_input | 0 - 397 | Índice demográfico |
| `lstat` | number_input | 1 - 38 | Estatus socioeconómico bajo (%) |

### Salidas

- **Precio estimado** en miles de USD
- **Segmento de precio**: Bajo (<$15k), Medio ($15-25k), Alto ($25-35k), Premium (>$35k)
- **Factores principales**: comparación de los valores ingresados vs el promedio del dataset

### Componentes técnicos

- **Framework**: Streamlit
- **Modelo**: Pipeline completo (ColumnTransformer + modelo ML) cargado desde `models/best_model.joblib`
- **Cache**: el modelo se carga una sola vez usando `@st.cache_resource`

## 🏗️ Estructura del código

````python
# Verificar que el modelo existe y se puede cargar
import joblib

model_path = "../../models/best_model.joblib"
model = joblib.load(model_path)

print("Modelo cargado correctamente")
print(f"\nPipeline:")
for step_name, step in model.named_steps.items():
    print(f"  {step_name}: {type(step).__name__}")
````

````python
# Prueba de predicción con valores promedio
import pandas as pd

sample = pd.DataFrame(
    [
        {
            "crim": 3.61,
            "zn": 11.36,
            "indus": 11.14,
            "chas": "0",
            "nox": 0.55,
            "rm": 6.28,
            "age": 68.57,
            "dis": 3.80,
            "rad": 4.0,
            "tax": 408.0,
            "ptratio": 18.46,
            "black": 356.67,
            "lstat": 12.65,
        }
    ]
)

prediction = model.predict(sample)[0]
print(f"Predicción con valores promedio: ${prediction:,.2f} miles USD")
print(f"\nEsto confirma que el modelo funciona correctamente para la demo.")
````
