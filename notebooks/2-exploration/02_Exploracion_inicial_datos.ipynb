## 🔍 Unificar la representación de valores nulos

Se verificará si existen diferentes representaciones de valores nulos en el dataset (NaN, cadenas vacías, valores como -999, "NA", "null", etc.) y se unificarán a `NaN` de pandas.

```python
# Conteo de valores nulos actuales (NaN nativos de pandas)
print("=== Valores nulos por columna ===")
null_counts = df.isnull().sum()
null_pct = (df.isnull().sum() / len(df) * 100).round(2)
null_summary = pd.DataFrame({"Nulos": null_counts, "% Nulos": null_pct})
print(null_summary)
print(f"\nTotal de celdas nulas: {df.isnull().sum().sum()}")
print(f"Porcentaje total: {(df.isnull().sum().sum() / df.size * 100):.2f}%")
```

```python
# Verificar si hay otros valores que representen nulos
# (cadenas vacías, -999, "NA", "null", "N/A", "?", etc.)
POSSIBLE_NULL_VALUES = ["", " ", "NA", "N/A", "na", "n/a", "null", "NULL", "None", "?", "-", "--", "-999", "999"]

print("=== Búsqueda de representaciones alternativas de nulos ===")
found_alt_nulls = False
for col in df.columns:
    for null_val in POSSIBLE_NULL_VALUES:
        count = (df[col].astype(str) == null_val).sum()
        if count > 0:
            print(f"  Columna '{col}': {count} valores con representación '{null_val}'")
            found_alt_nulls = True

if not found_alt_nulls:
    print("  No se encontraron representaciones alternativas de nulos.")
    print("  Todos los valores faltantes ya están representados como NaN.")
```

```python
# Verificar valores negativos inesperados que podrían ser nulos codificados
print("=== Valores negativos por columna ===")
for col in df.select_dtypes(include=[np.number]).columns:
    neg_count = (df[col] < 0).sum()
    if neg_count > 0:
        print(f"  {col}: {neg_count} valores negativos")

if all((df[col] >= 0).all() or df[col].isna().any() for col in df.select_dtypes(include=[np.number]).columns):
    print("  No se encontraron valores negativos inesperados.")
```

**Conclusión sobre nulos:** Todos los valores faltantes del dataset están representados de forma uniforme como `NaN` nativo de pandas. No se encontraron representaciones alternativas de nulos que requieran unificación.

## 🔧 Conversión de tipos de datos y corrección de inconsistencias

En la Tarea 1 se detectaron inconsistencias de escala en las columnas `crim`, `nox`, `rm` y `dis`, donde algunos valores están multiplicados por 1000 respecto a los rangos esperados del dataset original de Boston Housing. En esta sección se corrigen esos problemas y se asignan los tipos de datos correctos.

### Paso 1: Corrección de inconsistencias de escala

Las columnas afectadas y sus umbrales de detección son:

| Columna | Rango esperado | Umbral de detección | Valores afectados |
|---------|---------------|--------------------|-----------|
| `crim` | 0 - 89 | > 100 | Valores multiplicados por ×1000 |
| `nox` | 0.38 - 0.87 | > 1 | Valores multiplicados por ×1000 |
| `rm` | 3.5 - 8.8 | > 100 | Valores multiplicados por ×1000 |
| `dis` | 1.1 - 12.1 | > 100 | Valores multiplicados por ×1000 |

```python
# Crear copia para trabajar
df_clean = df.copy()

# Definir las columnas con problemas de escala y sus umbrales
SCALE_CORRECTIONS = {
    "crim": 100,
    "nox": 1,
    "rm": 100,
    "dis": 100,
}
SCALE_FACTOR = 1000

print("=== Corrección de escala ===")
for col, threshold in SCALE_CORRECTIONS.items():
    mask = df_clean[col] > threshold
    affected = mask.sum()
    print(f"\n{col} (umbral > {threshold}):")
    print(f"  Valores afectados: {affected}")
    print(f"  Antes  -> min: {df_clean[col].min():.4f}, max: {df_clean[col].max():.4f}")
    df_clean.loc[mask, col] = df_clean.loc[mask, col] / SCALE_FACTOR
    print(f"  Después -> min: {df_clean[col].min():.4f}, max: {df_clean[col].max():.4f}")
```

```python
# Verificar que los rangos corregidos son coherentes con el dataset original
EXPECTED_RANGES = {
    "crim": (0, 89),
    "zn": (0, 100),
    "indus": (0, 28),
    "chas": (0, 1),
    "nox": (0.3, 0.9),
    "rm": (3, 9),
    "age": (0, 100),
    "dis": (1, 13),
    "rad": (1, 24),
    "tax": (180, 720),
    "ptratio": (12, 22),
    "black": (0, 400),
    "lstat": (1, 40),
    "medv": (4, 51),
}

print("=== Verificación de rangos post-corrección ===")
for col, (expected_min, expected_max) in EXPECTED_RANGES.items():
    actual_min = df_clean[col].min()
    actual_max = df_clean[col].max()
    status = "✅" if actual_min >= expected_min * 0.5 and actual_max <= expected_max * 1.5 else "⚠️"
    print(f"  {status} {col}: [{actual_min:.2f}, {actual_max:.2f}] (esperado: [{expected_min}, {expected_max}])")
```

### Paso 2: Conversión de tipos de datos

Se asignan los tipos de datos correctos a cada columna según su naturaleza:

- **`ID`**: entero — identificador, no se usa en el modelo.
- **`chas`**: categórica binaria (0/1) — variable dummy del río Charles.
- **`rad`**: entero (ordinal) — índice de accesibilidad.
- **Resto**: float64 — variables numéricas continuas.

```python
print("=== Tipos de datos antes de conversión ===")
print(df_clean.dtypes)

# Convertir 'chas' a tipo categórico
df_clean["chas"] = df_clean["chas"].astype("Int64").astype("category")

# Convertir 'rad' a entero (nullable porque puede tener nulos)
df_clean["rad"] = df_clean["rad"].astype("Int64")

# Mantener 'ID' como entero
df_clean["ID"] = df_clean["ID"].astype("int64")

# Las demás columnas ya son float64, que es correcto
print("\n=== Tipos de datos después de conversión ===")
print(df_clean.dtypes)
```

```python
# Verificar que cada columna tiene un tipo uniforme
print("=== Verificación de uniformidad de tipos ===")
for col in df_clean.columns:
    dtype = df_clean[col].dtype
    non_null = df_clean[col].dropna()
    print(f"  {col}: {dtype} — {len(non_null)} valores no nulos")
```

### Paso 3: Revisión de duplicados

```python
# Verificar filas duplicadas (excluyendo ID)
cols_sin_id = [col for col in df_clean.columns if col != "ID"]
duplicados = df_clean.duplicated(subset=cols_sin_id, keep=False)
print(f"Filas duplicadas (excluyendo ID): {duplicados.sum()}")

if duplicados.sum() > 0:
    print("\nRegistros duplicados:")
    print(df_clean[duplicados].sort_values(cols_sin_id))
```

### Paso 4: Resumen del dataset limpio

```python
print("=" * 60)
print("RESUMEN DEL DATASET DESPUÉS DE LA EXPLORACIÓN INICIAL")
print("=" * 60)
print(f"\nDimensiones: {df_clean.shape[0]} filas x {df_clean.shape[1]} columnas")
print(f"Valores nulos totales: {df_clean.isnull().sum().sum()}")
print(f"\nCorrecciones aplicadas:")
print("  1. Escala corregida en: crim, nox, rm, dis (÷1000 donde correspondía)")
print("  2. chas convertida a tipo categórico (0/1)")
print("  3. rad convertida a entero (Int64, nullable)")
print(f"\nValores nulos restantes (a tratar en tareas posteriores):")
null_remaining = df_clean.isnull().sum()
print(null_remaining[null_remaining > 0])
```

## 💾 Almacenamiento del dataset en formato Parquet

Se almacena el dataset limpio en formato `.parquet` en la carpeta `data/02_intermediate/`. Este formato es más eficiente que CSV en términos de almacenamiento y velocidad de lectura, y preserva los tipos de datos.

```python
# Guardar en formato parquet
output_path = "../../data/02_intermediate/boston_housing_clean.parquet"
df_clean.to_parquet(output_path, index=False, engine="pyarrow")

# Verificar que se guardó correctamente
df_verify = pd.read_parquet(output_path)
print(f"Archivo guardado en: {output_path}")
print(f"Dimensiones verificadas: {df_verify.shape}")
print(f"Tipos preservados: {(df_verify.dtypes == df_clean.dtypes).all()}")
```

## 📊 Analysis of Results and Conclusions

### Resultados

La exploración inicial del dataset de Boston Housing reveló y corrigió los siguientes problemas:

1. **Valores nulos**: se identificaron 66 celdas nulas distribuidas en todas las columnas. La variable objetivo (`medv`) tiene 16 nulos (3.6%). Todos los nulos ya estaban representados como `NaN` de pandas, no se requirió unificación.

2. **Inconsistencias de escala**: se corrigieron 4 columnas (`crim`, `nox`, `rm`, `dis`) donde algunos valores estaban multiplicados por 1000. 3. **Tipos de datos**: se convirtió `chas` a tipo categórico (variable dummy binaria) y `rad` a entero nullable (`Int64`). Las demás columnas numéricas se mantienen como `float64`.

4. **Formato de almacenamiento**: el dataset limpio se guardó en formato `.parquet` que preserva los tipos de datos y es más eficiente.

- El principal problema de calidad del dataset eran las inconsistencias de escala, no los valores nulos.
- Los valores nulos no se imputaron en esta etapa ya que su tratamiento depende del análisis exploratorio (Tarea 3) y del feature engineering (Tarea 4).
