## 🔍 Entendimiento del Problema

A continuación se responden las preguntas requeridas para comprender el problema antes de iniciar el proceso de ciencia de datos.

### 1. ¿Cuál es el objetivo del problema?

El objetivo es **predecir el valor mediano de las viviendas ocupadas por sus propietarios** (`medv`, en miles de dólares) en distintas zonas de Boston, a partir de 13 atributos que describen características socioeconómicas y geográficas de cada zona (tasa de criminalidad, número promedio de habitaciones, distancia a centros de empleo, proporción de terreno residencial, entre otros).

Se trata de un problema de **regresión** donde la variable objetivo es continua.

### 2. ¿Cómo se usará su solución?

La solución puede utilizarse como una herramienta de estimación de precios de viviendas. Potenciales usos incluyen:

- **Agentes inmobiliarios**: estimar el precio de mercado de una vivienda en función de las características de la zona.
- **Compradores/vendedores**: obtener una referencia de precio antes de una transacción.
- **Entidades financieras**: evaluar el valor de una propiedad como garantía para créditos hipotecarios.
- **Urbanistas y gobierno local**: analizar qué factores inciden más en el valor de las viviendas para orientar políticas públicas.

En el contexto de este curso, la solución sirve como **prueba de concepto (POC)** para aplicar un proceso completo de ciencia de datos con enfoque MLOps.

### 3. ¿Cuáles son las soluciones actuales (si las hay)?

- **Avalúos manuales**: peritos inmobiliarios que visitan la propiedad y la valoran con base en su experiencia, comparables del mercado y tablas de referencia.
- **Modelos hedónicos de precios**: modelos de regresión lineal que descomponen el precio en función de los atributos de la vivienda y su entorno. Son ampliamente usados en economía inmobiliaria.
- **Plataformas inmobiliarias** (Zillow Zestimate, Redfin, etc.): utilizan modelos de machine learning a gran escala con datos actualizados del mercado.
- **Soluciones académicas sobre este mismo dataset**: el dataset de Boston Housing es uno de los más clásicos en ML. Se han publicado múltiples soluciones con regresión lineal, árboles de decisión, Random Forest, Gradient Boosting y redes neuronales.

### 4. ¿Cómo se debe enmarcar este problema (supervisado/no supervisado, en línea/fuera de línea, etc.)?

- **Tipo de aprendizaje**: Supervisado, ya que se dispone de una variable objetivo conocida (`medv`).
- **Tipo de tarea**: Regresión, puesto que la variable a predecir es continua (precio en miles de dólares).
- **Modalidad**: Fuera de línea (batch), ya que el dataset es estático y no se requiere actualización en tiempo real.
- **Salida**: Univariada — se predice una única variable numérica.

### 5. ¿Cómo se debe medir el desempeño o el rendimiento de la solución, una primera intuición?

Para un problema de regresión, las métricas candidatas son:

| Métrica | Descripción | Ventaja |
|---------|-------------|--------|
| **RMSE** (Root Mean Squared Error) | Raíz del error cuadrático medio | Penaliza errores grandes, mismas unidades que la variable objetivo (miles USD) |
| **MAE** (Mean Absolute Error) | Error absoluto medio | Más robusta a outliers, fácil de interpretar |
| **R²** (Coeficiente de determinación) | Proporción de varianza explicada | Valor entre 0 y 1, fácil de comunicar |
| **MAPE** (Mean Absolute Percentage Error) | Error porcentual absoluto medio | Interpretación relativa al precio |

**Primera intuición**: usar **RMSE** como métrica principal ya que penaliza los errores grandes de predicción (que en un contexto inmobiliario son los más costosos) y se expresa en las mismas unidades que el precio (miles de dólares). Complementar con **R²** para comunicar el rendimiento del modelo de forma intuitiva.

### 6. ¿La medida de desempeño está alineada con el objetivo del problema?

Sí. El RMSE captura directamente el error promedio en la predicción del precio y penaliza las desviaciones grandes, lo cual es coherente con el objetivo de estimar precios de viviendas de forma confiable. Un error grande en la estimación de un precio puede tener consecuencias financieras significativas, por lo que una métrica que penalice estos errores resulta adecuada.

El R² complementa indicando qué proporción de la variabilidad del precio es capturada por el modelo, lo cual es útil para evaluar si el modelo aporta valor frente a simplemente usar el promedio.

### 7. ¿Cuál sería el desempeño o rendimiento mínimo necesario para alcanzar el objetivo del problema?

Considerando que la desviación estándar del precio (`medv`) es aproximadamente **$9,220 USD**, un modelo útil debería tener:

- **RMSE < $5,000 USD** (menor que la mitad de la desviación estándar del target), lo que indicaría que el modelo predice mejor que simplemente estimar con la media.
- **R² > 0.70**, lo que significaría que el modelo explica al menos el 70% de la variabilidad de los precios.

Estos umbrales se definen como el mínimo aceptable para la POC. En un contexto de producción real, se esperaría un RMSE más bajo y un R² cercano a 0.85-0.90.

### 10. (Importante) ¿Cómo se puede resolver el problema manualmente?

Un experto inmobiliario podría estimar manualmente el precio de una vivienda en Boston usando reglas heurísticas como:

1. **Partir de un precio base** cercano a la mediana (~$20,600 USD).
2. **Ajustar según el número de habitaciones** (`rm`): a mayor número de habitaciones, mayor precio. Cada habitación adicional podría representar un incremento de ~$3,000-$5,000.
3. **Penalizar por criminalidad** (`crim`): zonas con alta tasa de criminalidad tendrían un descuento.
4. **Penalizar por proporción de población de estatus bajo** (`lstat`): a mayor porcentaje, menor precio.
5. **Considerar cercanía a centros de empleo** (`dis`): la distancia a centros laborales influye en el precio.
6. **Factor río Charles** (`chas`): las viviendas colindantes con el río tienen un precio premium.

Este enfoque manual tendría errores significativos pero serviría como **modelo heurístico base** para comparar con los modelos de ML.

```python
# Revisión de valores nulos por columna
print("=== Valores nulos por columna ===")
null_counts = df.isnull().sum()
null_pct = (df.isnull().sum() / len(df) * 100).round(2)
null_summary = pd.DataFrame({"Nulos": null_counts, "% Nulos": null_pct})
print(null_summary[null_summary["Nulos"] > 0])
```

```python
# Verificar inconsistencias de escala
# Según el dataset original de Boston, 'rm' debería estar entre ~3-9
# pero se observan valores como 6575 (posiblemente rm * 1000)

SCALE_THRESHOLD = 100
RM_UPPER_LIMIT = 10

print("=== Detección de inconsistencias de escala ===")
print("\nrm (esperado ~3-9 habitaciones):")
print(f"  Valores > {SCALE_THRESHOLD}: {(df['rm'] > SCALE_THRESHOLD).sum()} registros")
print(f"  Valores < {RM_UPPER_LIMIT}:  {(df['rm'] < RM_UPPER_LIMIT).sum()} registros")

print("\nnox (esperado ~0.3-0.9 partes por 10M):")
print(f"  Valores > 1:   {(df['nox'] > 1).sum()} registros")
print(f"  Valores <= 1:  {(df['nox'] <= 1).sum()} registros")

print("\ndis (esperado ~1-13):")
print(f"  Valores > {SCALE_THRESHOLD}: {(df['dis'] > SCALE_THRESHOLD).sum()} registros")
print(f"  Valores < {SCALE_THRESHOLD}: {(df['dis'] < SCALE_THRESHOLD).sum()} registros")
```

```python
# Verificar IDs - ¿hay registros faltantes del dataset original?
print("=== Análisis de IDs ===")
print(f"Registros presentes: {len(df)}")
print(f"Rango de ID: {df['ID'].min()} - {df['ID'].max()}")
ids_faltantes = set(range(1, df["ID"].max() + 1)) - set(df["ID"])
print(f"IDs faltantes: {len(ids_faltantes)}")
print(
    f"Esto sugiere que se eliminaron {len(ids_faltantes)} registros del dataset original."
)
```
